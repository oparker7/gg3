## Model Simulation and Behaviour

## Task 1.1

```python
import models
import numpy as np
import matplotlib.pyplot as plt
import datavis as dv


# construct a model from each of the two provided classes
ramp = models.RampModel(beta=0.5, sigma=0.2)
step = models.StepModel(m=500, r=10, x0=.2, Rh=50, isi_gamma_shape=None, Rl=None, dt=None)
```

## Ramp Model

```python
# decide simulation parameters
Ntrials_ramp = 5
T_ramp = 10

# generate simulation of ramp model

r_sim = ramp.simulate(Ntrials = Ntrials_ramp, T= T_ramp)

dv.raster_plot(r_sim[0])

dv.walk_plot(r_sim[1])
```

## Step Model

```python
Ntrials_step = 10
T_step = 1000


s_sim = step.simulate(Ntrials = Ntrials_step, T = T_step)

dv.raster_plot(s_sim[0])

dv.jump_hist(s_sim[1])
```
